## 2. Three ways to read the documentation

On the slides we looked up `np.linspace`. Help works the same way for
everything in Python, including the built-ins you already use. Look up the
built-in **`range`** the same three ways:

1. `help(range)`, which works in any Python session
2. `range?`, which works in Jupyter and IPython (`range??` tries to show the source, but
   built-ins are written in C, so there is none to show)
3. The online docs: search for "python range", or go to <https://docs.python.org/3/library/functions.html>

Then answer the questions below.

## 3. Read a traceback

The cell below fails on purpose. It reports the range of a few voltage
samples from the brain recording we use all course (you meet it properly in
the NumPy block). The samples were read from a text file. The error happens
inside a function, which is called from another function. Read the traceback
from the **bottom up**:

1. What type of error is it, and what does the message say?
2. In which function, and on which line, did it break?
3. Which line in the cell started the chain of calls?
4. Fix it in the empty cell underneath, **without changing either function**.

```python
def peak_to_peak(values):
    return max(values) - min(values)


def report(name, values):
    print(name, "range:", peak_to_peak(values))


samples = ["112.5", "-37.5", "84.4"]  # as read from the text file
report("LFP, channel 100 (µV)", samples)
```

1. A `TypeError`: `unsupported operand type(s) for -: 'str' and 'str'`.
2. In `peak_to_peak`, on the line `return max(values) - min(values)`.
3. The last line, `report("LFP, channel 100 (µV)", samples)`, which calls `report`,
   which calls `peak_to_peak`.
4. The samples are strings, not numbers. `max` and `min` still run on strings
   (they compare them alphabetically), but you cannot subtract one string from
   another. Convert the samples to numbers first.

```python
report("LFP, channel 100 (µV)", [112.5, -37.5, 84.4])

# or convert the strings you read from the file
report("LFP, channel 100 (µV)", [float(s) for s in samples])
```
